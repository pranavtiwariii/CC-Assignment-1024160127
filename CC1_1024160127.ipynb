```python
roll_no = "1024160127"

l = [int(digit) * 10 for digit in roll_no]

print("original list:", l)

l.append(250)
print("after append:", l)

l.insert(2, 350)
print("after insert:", l)

l.remove(250)
print("after remove:", l)

l.pop(2)
print("after pop:", l)

l.sort()
print("ascending order:", l)

l.sort(reverse=True)
print("descending order:", l)

print("first three elements:", l[:3])
print("last three elements:", l[-3:])

average = sum(l) / len(l)
greater_than_average = [x for x in l if x > average]

print("average:", average)
print("elements greater than average:", greater_than_average)
```

```python
scores = tuple(l[:8])

print("scores tuple:", scores)

highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("highest score:", highest)
print("index of highest score:", highest_index)
print("lowest score:", lowest)
print("number of times lowest appears:", lowest_count)

reversed_scores = list(reversed(scores))

print("reversed tuple as list:", reversed_scores)

user_score = int(input("enter a score: "))

if user_score in scores:
    print("first occurrence index:", scores.index(user_score))
else:
    print("score is not present in the tuple.")

try:
    scores[0] = 100
except TypeError as e:
    print("error:", e)

first_score, second_score, *remaining_scores = scores

print("first score:", first_score)
print("second score:", second_score)
print("remaining scores:", remaining_scores)
```

```python
import random

random.seed(1024160127)

numbers = [random.randint(100, 900) for _ in range(100)]

print("random numbers:")
print(numbers)

odd_numbers = [x for x in numbers if x % 2 != 0]

print("odd numbers:")
print(odd_numbers)
print("number of odd numbers:", len(odd_numbers))

even_numbers = [x for x in numbers if x % 2 == 0]

print("even numbers:")
print(even_numbers)
print("number of even numbers:", len(even_numbers))

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

prime_numbers = [x for x in numbers if is_prime(x)]

print("prime numbers:")
print(prime_numbers)
print("number of prime numbers:", len(prime_numbers))

frequency = {}

for x in numbers:
    frequency[x] = frequency.get(x, 0) + 1

most_frequent = max(frequency, key=frequency.get)

print("most frequent number:", most_frequent)
print("number of times it occurs:", frequency[most_frequent])
```

```python
roll_no = "1024160127"

digits = [int(digit) for digit in roll_no]

a = {digit * 7 for digit in digits}
b = {digit * 9 for digit in digits}

print("set a:", a)
print("set b:", b)

union = a.union(b)
print("union:", union)

intersection = a.intersection(b)
print("intersection:", intersection)

a_minus_b = a.difference(b)
b_minus_a = b.difference(a)

print("a - b:", a_minus_b)
print("b - a:", b_minus_a)

symmetric_diff = a.symmetric_difference(b)
print("symmetric difference:", symmetric_diff)

print("is a subset of b:", a.issubset(b))
print("is b superset of a:", b.issuperset(a))

x = int(input("enter a value to discard from a: "))

a.discard(x)

print("set a after discard:", a)
```

```python
my_dict = {
    "name": "pranav",
    "roll_no": "1024160127",
    "branch": "cse",
    "age": 20,
    "city": "ncr"
}

print("original dictionary:")
print(my_dict)

my_dict["location"] = my_dict.pop("city")

print("after renaming city to location:")
print(my_dict)

my_dict["cgpa"] = 8.4

print("after adding cgpa:")
print(my_dict)

my_dict["age"] += 1

print("after increasing age:")
print(my_dict)

dict_pop = my_dict.copy()

removed_branch = dict_pop.pop("branch")

print("dictionary after pop:")
print(dict_pop)
print("value returned by pop:", removed_branch)

dict_del = my_dict.copy()

del dict_del["branch"]

print("dictionary after del:")
print(dict_del)

print("key-value pairs:")

for key, value in my_dict.items():
    print(f"{key} → {value}")

if "email" in my_dict:
    print("email:", my_dict["email"])
else:
    print("email key does not exist.")

friend_dict = {
    "name": "tiwari",
    "roll_no": "1024160128",
    "branch": "ece",
    "age": 19,
    "city": "ggn"
}

merged_dict = {**my_dict, **friend_dict}

print("merged dictionary:")
print(merged_dict)

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("dictionary containing only string values:")
print(string_values)
```
