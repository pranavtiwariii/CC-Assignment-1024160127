```python
import io
raw = '''Employee_ID,Name,Department,Age,Salary,Years_of_Experience,Joining_Date,Gender,Bonus,Rating
101,Alice,HR,29,50000,4,2020-03-15,Female,5000,4.5
102,Bob,IT,34,70000,8,2017-07-19,Male,7000,4.0
103,Charlie,IT,41,65000,10,2013-06-01,Male,6000,3.8
104,Diana,Marketing,28,55000,3,2021-02-10,Female,4500,4.7
105,Edward,Sales,38,60000,12,2010-11-25,Male,5000,3.5
'''
with open("employees.csv", "w") as f:
    f.write(raw)

emp = pd.read_csv("employees.csv")

print(emp.shape)

emp.info()

print(emp.describe())

print(emp.head(5))
print(emp.tail(3))

print(emp["Salary"].mean())
print(emp["Bonus"].sum())
print(emp["Age"].min())
print(emp["Rating"].max())

print(emp.sort_values("Salary", ascending=False))

emp["Performance"] = pd.cut(emp["Rating"], bins=[0, 4.0, 4.5, 10], labels=["Average", "Good", "Excellent"], right=False)
print(emp)

print(emp.isnull().sum())

emp = emp.rename(columns={"Employee_ID": "ID"})
print(emp.columns)

print(emp[emp["Years_of_Experience"] > 5])
print(emp[emp["Department"] == "IT"])

emp["Tax"] = emp["Salary"] * 0.10
print(emp)

emp.to_csv("employees_modified.csv", index=False)
```
