```python
import numpy as np

a = np.array([4, 8, 12, 16, 20])
print(a + 2)
print(a * 3)
print(a / 2)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print(arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

for z in (x, y):
    vals, counts = np.unique(z, return_counts=True)
    for v in vals[counts == counts.max()]:
        print(v, np.where(z == v)[0])
    print()
```

```python
daiwik = np.linspace(10, 100, 25)
print(daiwik)
print(daiwik.ndim)
print(daiwik.shape)
print(daiwik.size)
print(daiwik.dtype)
print(daiwik.nbytes)

t = daiwik.reshape(25, 1)
print(t)
print(t.shape)

print(daiwik.T.shape)
```

```python
ucs420_daiwik = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print(ucs420_daiwik)
print(np.mean(ucs420_daiwik))
print(np.median(ucs420_daiwik))
print(np.max(ucs420_daiwik))
print(np.min(ucs420_daiwik))
print(np.unique(ucs420_daiwik))

reshaped_ucs420_daiwik = ucs420_daiwik.reshape(4, 3)
print(reshaped_ucs420_daiwik)

resized_ucs420_daiwik = np.resize(ucs420_daiwik, (2, 3))
print(resized_ucs420_daiwik)
```
