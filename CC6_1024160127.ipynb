```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected = temperature + 2
print(corrected)

fahrenheit = corrected * 9 / 5 + 32
print(fahrenheit)

above = temperature[temperature > 32]
print(above)

print(np.sum(temperature > 32))

print("vectorization applies an operation to the whole array at once in compiled code, and boolean indexing filters values with a condition in one step, so both avoid the slow python loop and need less code")
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]])

print(steps.sum())
print(steps.mean())
print(steps.max(), steps.min())
print(steps.sum(axis=0))
print(steps.sum(axis=1))
pos = np.unravel_index(steps.argmax(), steps.shape)
print(steps.argmax(), pos)
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

print(X.shape)
print(X.ndim)

print(X.T)
print("the transpose swaps rows and columns, so each row now holds one feature across all users")

xtx = X.T @ X
print(xtx)

xtx_inv = np.linalg.inv(xtx)
print(xtx_inv)

beta = xtx_inv @ X.T @ y
print(beta)
print("each coefficient is the change in predicted assistance score for one unit increase in sleep hours, activity level and stress level respectively, keeping the other two fixed")

new_user = np.array([5, 40, 7])
print(new_user @ beta)
```
